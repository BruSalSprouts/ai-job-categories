### 4.2 Checking the duplicate values ###

For this one, I'm aware that duplicate values by column exist. What I'm curious to find are the unique values in different columns, and if there's any duplicate rows in this dataframe

```python
# This is the code to check the number of unique values in each column of the dataset.
print(f"The total number of values in this dataset is   {len(organized_job_data)}")
columns_with_small_unique_values = []
for column in organized_job_data.columns:
    sum_unique = organized_job_data[column].nunique()
    print(f"Number of unique values in {column}: {sum_unique}")
    if sum_unique < 20:  # Adjust the threshold as needed - Chose 38 as everything at or above it will take too much space in the notebook and will be hard to read.
        columns_with_small_unique_values.append(column)

print(f"Columns with small amount of unique values: {columns_with_small_unique_values}")
```

```python
unique_values_text = ""
for c in columns_with_small_unique_values:
    # print(f"Values in column {c}:")
    unique_values_text += f"Values in column {c}:\n"
    for val in organized_job_data[c].unique():
        count = len(organized_job_data[organized_job_data[c] == val])
        # print(f"    {val}: {count}")
        unique_values_text += f"    {val}: {count}\n"
# print(unique_values_text)
# Saves the unique values text to a file
with open("../data/processed/unique_values.txt", "w") as f:
    f.write(unique_values_text) 
```

#### 4.4.1 Converting CAD to USD ####
Now we need to convert the Canadian dollars in this to United States dollars

According to https://www.exchange-rates.org/exchange-rate-history/cad-usd-2025, the average exchange rate of CAD to USD in 2025 was 0.7158

```python
# Now we need to convert the Canadian dollars in this american_cleaned_job_data dataset to USD.
# First let's get all the unique values in the 'Salary Range' column to see what we need to convert.
salary_ranges_unconverted = american_cleaned_job_data['Salary Range'].tolist()
# salary_ranges_unconverted = american_cleaned_job_data['Salary Range'].unique().tolist()

canadian_salary_ranges_unconverted = []
for salary in salary_ranges_unconverted:
    if 'C' in salary:
        canadian_salary_ranges_unconverted.append(salary)
print(canadian_salary_ranges_unconverted[:10])  # Print the first 10 values to check
```

```python
# We'll split each string into a list of 2 values, separated by ' - '. 
list_of_salary_ranges = []
print(f"The beginning version: {canadian_salary_ranges_unconverted[:10]}")
temp = ""
for salary in canadian_salary_ranges_unconverted:
    temp = salary.split(' - ')
    temp[0] = temp[0][2:] # Removes the first 2 characters, which are the currency symbols
    temp[1] = temp[1][2:] # Removes the first 2 characters
    list_of_salary_ranges.append(temp)

# Each value will be converted to a numerical value after removing the first two characters, which are the currency symbols. Then we'll multiply the values by the conversion rate to convert them to USD. 
for salary_range in list_of_salary_ranges:
    salary_range[0] = str(round(int(salary_range[0].replace(',', '')) / 0.7158)) # Converts the first value to USD
    salary_range[0] = "{:,.0f}".format(float(salary_range[0])) 
    salary_range[1] = str(round(int(salary_range[1].replace(',', '')) / 0.7158)) # Converts the second value to USD
    salary_range[1] = "{:,.0f}".format(float(salary_range[1]))

# Finally, we'll join the two values back together with ' - ' and replace the original value in the 'Salary Range' column with the converted value.
canadian_salary_ranges_converted = []
for i, salary_range in enumerate(list_of_salary_ranges):
    converted_salary_range = '$' + salary_range[0] + ' - $' + salary_range[1]
    canadian_salary_ranges_converted.append(converted_salary_range)
    
print(f"The ending version: {canadian_salary_ranges_converted[:10]}")
```

```python
# Now we'll replace the original values in the 'Salary Range' column with the converted values. 
salary_ranges_converted = salary_ranges_unconverted
index_c = 0
for salary in salary_ranges_converted:
    if 'C' in salary:
        index_s = salary_ranges_unconverted.index(salary)
        salary_ranges_converted[index_s] = canadian_salary_ranges_converted[index_c]
        index_c += 1
print(salary_ranges_converted[:10])
print(pd.Series(salary_ranges_unconverted).unique()[:10])
print(len(salary_ranges_unconverted))
```
