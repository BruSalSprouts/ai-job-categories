### 4.2 Checking the duplicate values ###

For this one, I'm aware that duplicate values by column exist. What I'm curious to find are the unique values in different columns, and if there's any duplicate rows in this dataframe

```python
# This is the code to check the number of unique values in each column of the dataset.
print(f"The total number of values in this dataset is   {len(organized_job_data)}")
columns_with_small_unique_values = []
for column in organized_job_data.columns:
    sum_unique = organized_job_data[column].nunique()
    print(f"Number of unique values in {column}: {sum_unique}")
    if sum_unique < 20:  # Adjust the threshold as needed - Chose 38 as everything at or above it will take too much space in the notebook and will be hard to read.
        columns_with_small_unique_values.append(column)

print(f"Columns with small amount of unique values: {columns_with_small_unique_values}")
```

```python
unique_values_text = ""
for c in columns_with_small_unique_values:
    # print(f"Values in column {c}:")
    unique_values_text += f"Values in column {c}:\n"
    for val in organized_job_data[c].unique():
        count = len(organized_job_data[organized_job_data[c] == val])
        # print(f"    {val}: {count}")
        unique_values_text += f"    {val}: {count}\n"
# print(unique_values_text)
# Saves the unique values text to a file
with open("../data/processed/unique_values.txt", "w") as f:
    f.write(unique_values_text) 
```

I saved the text to a file since it's a lot to process at once, and also because it lets me read through it later

### 4.3 Checking for Outliers ### 

If we stick with just numerical data types, then we only have 3 categories: Years of Experience Required, Company Rating, and Number of Applicants.

```python
organized_job_data.select_dtypes(include=['int64', 'float64']).head()
```

So an idea I had is to list the number of values in the unique categories to try to find outliers there.
But first, let's see if there's outliers in the numeric columns.

We'll start by having the list of category names with numeric data types, and the list of the other cateogry names from columns_with_small_unique_values

```python
print(columns_with_small_unique_values)
```

```python
numeric_type_columns = organized_job_data.select_dtypes(include=['int64', 'float64']).columns
numeric_type_columns = numeric_type_columns.tolist()
other_type_columns = [col for col in columns_with_small_unique_values if col not in numeric_type_columns]
```

```python
# This will help try to find outliers in the columns that aren't categorical. We'll use different plots to visualize the distribution of values in each column and identify any potential outliers.
for column in numeric_type_columns:
    plt.figure(figsize=(10, 5))
    sns.histplot(x=organized_job_data[column])
    plt.title(f'Distribution of {column}')
    plt.show()
```

I didn't really notice any outliers here, so we'll try the others instead

```python
# Help me plotting the occurrence of each value in the categorical columns. We'll use bar plots to visualize the frequency of each unique value in the categorical columns and identify any potential outliers or imbalances in the data.
for column in other_type_columns:
    plt.figure(figsize=(10, 5))
    sns.countplot(y=organized_job_data[column], order=organized_job_data[column].value_counts().index)
    plt.title(f'Frequency of {column}')
    plt.show()
```

```python
print(other_type_columns)
cleaned_job_data = organized_job_data.copy() # This is to use later / keep a copy that will be modified. We'll make this organized_job_data later in the notebook to keep the original dataset intact for reference and comparison.
print(len(cleaned_job_data))
```

These code blocks will be used to see the outliers in the categorical columns

```python
# For frequency of company industry, we'll check the values that have 'Marketing & Creative Tech', 'Information Services', and ''Venture Capital & Startups' 
cleaned_job_data['Company Industry'].value_counts()   
```

So the three industries listed are outliers, and we can probably get rid of them or ignore them since having such little data with them could cause issues down the line

```python
# Let's remove values from company industry that include 'Marketing & Creative Tech', 'Information Services', and 'Venture Capital & Startups' 
cleaned_job_data = cleaned_job_data[~cleaned_job_data['Company Industry'].isin(['Marketing & Creative Tech', 'Information Services', 'Venture Capital & Startups'])]
cleaned_job_data['Company Industry'].value_counts()   
```

```python
# For frequency of company size, we'll check the values that have '1,000-5,000 employees' and '10-25 employees' and compare it to the '10,000+ employees' value
cleaned_job_data['Company Size'].value_counts() 
```

So the biggest outlier would be 1000-5000 employees, I do think it would be worth relabeling '1,000-5,000 employees' to '1,001-5,000 employees', since it fits better

```python
# Let's relabel '1,000-5,000 employees' to '1,001-5,000 employees'
cleaned_job_data['Company Size'] = cleaned_job_data['Company Size'].replace('1,000-5,000 employees', '1,001-5,000 employees')
cleaned_job_data['Company Size'].value_counts()
```

```python
# Here we'll compare the frequency of all the Employment Type values in the dataset to see if there are any imbalances or outliers.
cleaned_job_data['Employment Type'].value_counts()
```

While it'd be easy to say Full-time is an outlier here due to having such a higher amount of values than the others, it's not realy an outlier with context, so we can keep it in.

```python
# For frequency of Education Requirements, we'll check the all the values in the dataset to see if there are any imbalances or outliers.
cleaned_job_data['Education Requirements'].value_counts()
```

The outliers here are really just other ways of listing them with the others. It'd really be more beneficial to relabel them as the others

```python
# We'll relabel some of the values in Education Requirements to make them more consistent 

# temp_organized_job_data = organized_job_data.copy() # This was used to keep the values consistent, but wascommented out later to keep the original values in the dataset.

# First, Master's or PhD will go to the highest value, Master's in Computer Science, Machine Learning, or related field since the value won't cause much of a difference
cleaned_job_data['Education Requirements'] = cleaned_job_data['Education Requirements'].replace(['Master\'s or PhD'], 'Master\'s in Computer Science, Machine Learning, or related field')
# Next, Bachelor's in Software Engineering or equivalent experienc will go to Bachelor's in Computer Science or related field
cleaned_job_data['Education Requirements'] = cleaned_job_data['Education Requirements'].replace(['Bachelor\'s in Software Engineering or equivalent experience'], 'Bachelor\'s in Computer Science or related field')
# Next, PhD in NLP, CS, or Linguistics will go to PhD in Machine Learning, Computer Science, Statistics, or related quantitative field
cleaned_job_data['Education Requirements'] = cleaned_job_data['Education Requirements'].replace(['PhD in NLP, CS, or Linguistics'], 'PhD in Machine Learning, Computer Science, Statistics, or related quantitative field')
# Next, CS Bachelor's or equivalent experience will go to Bachelor's in Computer Science or related field
cleaned_job_data['Education Requirements'] = cleaned_job_data['Education Requirements'].replace(['CS Bachelor\'s or equivalent experience'], 'Bachelor\'s in Computer Science or related field')

# Now let's see the value counts again!
cleaned_job_data['Education Requirements'].value_counts()
```
