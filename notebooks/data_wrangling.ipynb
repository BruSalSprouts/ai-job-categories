### 4.2 Checking the duplicate values ###

For this one, I'm aware that duplicate values by column exist. What I'm curious to find are the unique values in different columns, and if there's any duplicate rows in this dataframe

```python
# This is the code to check the number of unique values in each column of the dataset.
print(f"The total number of values in this dataset is   {len(organized_job_data)}")
columns_with_small_unique_values = []
for column in organized_job_data.columns:
    sum_unique = organized_job_data[column].nunique()
    print(f"Number of unique values in {column}: {sum_unique}")
    if sum_unique < 20:  # Adjust the threshold as needed - Chose 38 as everything at or above it will take too much space in the notebook and will be hard to read.
        columns_with_small_unique_values.append(column)

print(f"Columns with small amount of unique values: {columns_with_small_unique_values}")
```

```python
unique_values_text = ""
for c in columns_with_small_unique_values:
    # print(f"Values in column {c}:")
    unique_values_text += f"Values in column {c}:\n"
    for val in organized_job_data[c].unique():
        count = len(organized_job_data[organized_job_data[c] == val])
        # print(f"    {val}: {count}")
        unique_values_text += f"    {val}: {count}\n"
# print(unique_values_text)
# Saves the unique values text to a file
with open("../data/processed/unique_values.txt", "w") as f:
    f.write(unique_values_text) 
```

I saved the text to a file since it's a lot to process at once, and also because it lets me read through it later
