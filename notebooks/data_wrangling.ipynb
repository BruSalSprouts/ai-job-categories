## 3 - Loading the Data ##

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("m0sm71/ai-jobs-dataset-2026")

print("Path to dataset files:", path)
```

### 3.1 Saving files to correct location ###

```python
# Save the file located in path to folder in data/raw
data_file = os.path.join("data", "raw", "ai_jobs_dataset_2026.csv")
```

```python
import shutil

# Copies the CSV files from kaggle cache to the data/raw folder
csv_files = ["ai_jobs_dataset_2026.csv", "raw_job_postings.csv"]
cwd = os.getcwd()
# Removes the last 9 characters from the path (notebooks) to get the parent directory we need as the destination
parent_dir = cwd[:-9]

for file in csv_files:
    source = os.path.join(path, file)
    destination = os.path.join(parent_dir, "data", "raw")
    shutil.copy(source, destination)
    print(f"Copied {file} to {destination}")
```
